## golde aggregation

```python
import sys, os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..", "src")))
from cdc_pipeline.config import SILVER_TABLE, GOLD_TABLE
from cdc_pipeline.transforms import  aggregate_by_status

silver_df = spark.read.table(SILVER_TABLE)
gold_df = aggregate_by_status(silver_df)
gold_df.write.mode("overwrite").saveAsTable(GOLD_TABLE)

```
