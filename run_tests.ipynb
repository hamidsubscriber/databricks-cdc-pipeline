```python
import shutil
import os

repo_path  = os.getcwd()
test_run_path = "/tmp/test_run_copy"

if os.path.exists(test_run_path):
       shutil.rmtree(test_run_path)
shutil.copytree(repo_path, test_run_path)

print(f"Copied repo to writable path: {test_run_path}")
```

```python
import sys, pytest
import os
sys.path.insert(0, os.path.join(test_run_path, "src"))
pytest.main(["-v", os.path.join(test_run_path, "tests")])
```
